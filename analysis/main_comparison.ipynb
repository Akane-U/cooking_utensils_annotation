# mainデータのアノテーター間比較

```python
import itertools
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib
import json
```

```python
with open('../data/recipe_100.json', 'r', encoding='utf-8') as f:
    recipe = json.load(f)

recipe
```

```python
PATH = "../outputs/"
ANNOTATORS = ["A", "B", "C"]

raw_data = {}
for ann in ANNOTATORS:
    with open(PATH + f"{ann}_main_reviewed.json", encoding="utf-8") as f:
        raw_data[ann] = json.load(f)

with open(PATH + "main_annotated.json", encoding="utf-8") as f:
        raw_data["admin"] = json.load(f)

n_recipes = len(raw_data["A"])
titles = [r["title"] for r in raw_data["A"]]
for ann in ANNOTATORS:
    assert len(raw_data[ann]) == n_recipes
    assert [r["title"] for r in raw_data[ann]] == titles

titles
```

```python
# アノテーション進捗の確認

def per_recipe_counts(data):
    """レシピごとの (state数, interaction数) を返す（step_after=0の材料一覧は除く）"""
    counts = []
    for recipe in data:
        n_states, n_inter = 0, 0
        for step in recipe["world_state_list"]:
            if step["step_after"] == 0:
                continue
            for state in step["state_list"]:
                n_states += 1
                n_inter += len(state["utensil_interactions_list"])
        counts.append((n_states, n_inter))
    return counts

admin_counts = per_recipe_counts(raw_data["admin"])

incomplete_recipes = {}
for ann in ANNOTATORS:
    ann_counts = per_recipe_counts(raw_data[ann])
    incomplete = [
        i for i, ((a_s, _), (b_s, _)) in enumerate(zip(admin_counts, ann_counts))
        if a_s > 0 and b_s == 0
    ]
    incomplete_recipes[ann] = incomplete
    print(f"{ann}: 未着手レシピ {len(incomplete)}/{n_recipes} 件")
    if incomplete:
        print("  ->", [titles[i] for i in incomplete])
```

```python
# 例題（練習用レシピ）は本番のミス集計から除外
EXAMPLE_IDX = [i for i, t in enumerate(titles) if t.startswith("例題")]
print("除外する例題:", [titles[i] for i in EXAMPLE_IDX])
```

## `*_check` が true のもの（=ミス有）を集計・分析

```python
def collect_checks(obj, annotator, rows):
    """末尾が '_check' のフィールドを全て収集する"""
    if isinstance(obj, dict):
        for k, v in obj.items():
            if k.endswith("_check"):
                rows.append({"annotator": annotator, "check_field": k, "value": bool(v)})
            collect_checks(v, annotator, rows)
    elif isinstance(obj, list):
        for item in obj:
            collect_checks(item, annotator, rows)

check_rows = []
for ann in ANNOTATORS:
    for ann_recipe in raw_data[ann]:
        collect_checks(ann_recipe, ann, check_rows)

check_df = pd.DataFrame(check_rows)

check_summary = (
    check_df.groupby(["check_field", "annotator"])["value"]
    .agg(true_count="sum", total="count")
    .reset_index()
)

check_table = check_summary.pivot(index="check_field", columns="annotator", values=["true_count", "total"])
check_table["true_count", "total"] = check_table["true_count"].sum(axis=1)
check_table["total", "total"] = check_table["total"].sum(axis=1)
check_table = check_table.reindex(columns=ANNOTATORS + ["total"], level=1)
check_table
```

```python
# 未着手レシピ・例題・`step_after=0` を除いて再集計
# -> `step_after=0` の state は `name_check` / `missing_interaction_check` の付与有無がアノテーターで揃っていない（スクリプトミス）ため除外

def collect_checks_filtered(recipe, annotator, rows):
    """step_after=0 を除いた state / interaction / step の '_check' フィールドを収集する"""
    for step in recipe["world_state_list"]:
        if step["step_after"] == 0:
            continue
        for k, v in step.items():
            if k.endswith("_check"):
                rows.append({"annotator": annotator, "check_field": k, "value": bool(v)})
        for state in step["state_list"]:
            for k, v in state.items():
                if k.endswith("_check"):
                    rows.append({"annotator": annotator, "check_field": k, "value": bool(v)})
            for inter in state["utensil_interactions_list"]:
                for k, v in inter.items():
                    if k.endswith("_check"):
                        rows.append({"annotator": annotator, "check_field": k, "value": bool(v)})

check_rows_v2 = []
for ann in ANNOTATORS:
    for i, ann_recipe in enumerate(raw_data[ann]):
        if i in incomplete_recipes[ann] or i in EXAMPLE_IDX:
            continue
        collect_checks_filtered(ann_recipe, ann, check_rows_v2)

check_df_v2 = pd.DataFrame(check_rows_v2)

check_summary_v2 = (
    check_df_v2.groupby(["check_field", "annotator"])["value"]
    .agg(true_count="sum", total="count")
    .reset_index()
)

check_table_v2 = check_summary_v2.pivot(index="check_field", columns="annotator", values=["true_count", "total"])
check_table_v2["true_count", "total"] = check_table_v2["true_count"].sum(axis=1)
check_table_v2["total", "total"] = check_table_v2["total"].sum(axis=1)
check_table_v2 = check_table_v2.reindex(columns=ANNOTATORS + ["total"], level=1)
check_table_v2
```

```python
# 正解率 = 1 - (true_count / total) として算出

accuracy_summary = check_summary_v2.copy()

# アノテーターごとの合計行を追加してから正解率を計算する
totals_by_field = (
    accuracy_summary.groupby("check_field")[["true_count", "total"]]
    .sum()
    .reset_index()
    .assign(annotator="total")
)
accuracy_summary = pd.concat([accuracy_summary, totals_by_field], ignore_index=True)
accuracy_summary["accuracy_rate"] = 1 - accuracy_summary["true_count"] / accuracy_summary["total"]

accuracy_table = accuracy_summary.pivot(index="check_field", columns="annotator", values="accuracy_rate")
accuracy_table = accuracy_table.reindex(columns=ANNOTATORS + ["total"])
(accuracy_table * 100).round(1)
```

### tools / vessels 以外のミス部分とコメントを確認

```python
def extract_non_tools_vessel_issues(raw_data, incomplete_recipes, example_idx, annotators):
    """tools_check / vessel_check 以外で check=true のものを、対応するコメントとともに収集する"""
    rows = []
    for ann in annotators:
        for i, recipe in enumerate(raw_data[ann]):
            if i in incomplete_recipes[ann] or i in example_idx:
                continue
            title = recipe["title"]
            for step in recipe["world_state_list"]:
                if step["step_after"] == 0:
                    continue

                if step.get("missing_state_check"):
                    comment = " / ".join(
                        step.get(c, "") for c in [
                            "missing_state_name_comment",
                            "missing_state_source_comment",
                            "missing_state_vessel_comment",
                            "missing_state_tools_comment",
                        ] if step.get(c)
                    )
                    rows.append({
                        "annotator": ann,
                        "check_field": "missing_state_check",
                        "title": title,
                        "step_after": step["step_after"],
                        "state_id": "",
                        "state_name": "",
                        "comment": comment,
                    })

                for state in step["state_list"]:
                    if state.get("name_check"):
                        rows.append({
                            "annotator": ann,
                            "check_field": "name_check",
                            "title": title,
                            "step_after": step["step_after"],
                            "state_id": state["id"],
                            "state_name": state["name"],
                            "comment": state.get("name_comment", ""),
                        })

                    if state.get("missing_interaction_check"):
                        comment = " / ".join(
                            state.get(c, "") for c in [
                                "missing_source_comment",
                                "missing_vessel_comment",
                                "missing_tools_comment",
                            ] if state.get(c)
                        )
                        rows.append({
                            "annotator": ann,
                            "check_field": "missing_interaction_check",
                            "title": title,
                            "step_after": step["step_after"],
                            "state_id": state["id"],
                            "state_name": state["name"],
                            "comment": comment,
                        })

                    for inter in state["utensil_interactions_list"]:
                        if inter.get("source_state_id_check"):
                            rows.append({
                                "annotator": ann,
                                "check_field": "source_state_id_check",
                                "title": title,
                                "step_after": step["step_after"],
                                "state_id": state["id"],
                                "state_name": state["name"],
                                "comment": inter.get("source_state_id_comment", ""),
                            })
    return pd.DataFrame(rows)

issues_df = extract_non_tools_vessel_issues(raw_data, incomplete_recipes, EXAMPLE_IDX, ANNOTATORS)
issues_df
```

### tools / vessels のミス部分とコメントを確認

```python
def extract_tools_vessel_issues(raw_data, incomplete_recipes, example_idx, annotators):
    """tools_check / vessel_check で check=true のものを、対応するコメントとともに収集する"""
    rows = []
    for ann in annotators:
        for i, recipe in enumerate(raw_data[ann]):
            if i in incomplete_recipes[ann] or i in example_idx:
                continue
            title = recipe["title"]
            for step in recipe["world_state_list"]:
                if step["step_after"] == 0:
                    continue

                for state in step["state_list"]:
                    for inter in state["utensil_interactions_list"]:
                        if inter.get("tools_check"):
                            rows.append({
                                "annotator": ann,
                                "check_field": "tools_check",
                                "title": title,
                                "step_after": step["step_after"],
                                "state_id": state["id"],
                                "state_name": state["name"],
                                "value": inter.get("tools"),
                                "comment": inter.get("tools_comment", ""),
                            })
                        if inter.get("vessel_check"):
                            rows.append({
                                "annotator": ann,
                                "check_field": "vessel_check",
                                "title": title,
                                "step_after": step["step_after"],
                                "state_id": state["id"],
                                "state_name": state["name"],
                                "value": inter.get("vessel"),
                                "comment": inter.get("vessel_comment", ""),
                            })
    return pd.DataFrame(rows)

issues_tools_vessel_df = extract_tools_vessel_issues(raw_data, incomplete_recipes, EXAMPLE_IDX, ANNOTATORS)
print(f"tools / vessels ミス検出レシピ数：{issues_tools_vessel_df['title'].nunique()}/{n_recipes - len(EXAMPLE_IDX)} 件")
issues_tools_vessel_df
```
